### Week 9: Assignment 9 - Color & Accessibility

I am using a chart from Assignment 2 where I used the "tips" seaborn dataset and looked at Average Tips by Gender.

```python
import seaborn as sns
import matplotlib.pyplot as plt
tips = sns.load_dataset("tips")

print(tips.head())
```

#### Original Chart

```python
sns.barplot(data=tips, x="day", 
            y="tip", 
            hue="sex", 
            palette="bright")
plt.title("Average Tips by Gender")
```

#### Improved Chart

```python
colors = ["#377eb8", "#f781bf"]

sns.barplot(data=tips, x="day", 
            y="tip", 
            hue="sex", 
            palette=colors)
plt.title("Average Tips by Gender")
```

#### Reflections

In my original visualization, I used Seaborn’s “bright” palette, which included blue and orange. This combination was visually unappealing and lacked sufficient contrast for some viewers with color vision deficiencies, particularly those with blue-yellow color blindness. To improve accessibility, I replaced the colors with a colorblind-safe palette: deep blue (#377eb8) and vivid pink (#f781bf). I chose these colors not only for accessibility but also because they are commonly associated with males and females, which helps make the chart’s categories more immediately intuitive for most audiences.\
This redesign improves clarity and inclusivity by making the two categories easier to distinguish, even for viewers with color vision deficiencies. The colors also maintain high contrast when printed in grayscale or viewed on low-quality screens. I verified the palette using the Coblis color blindness simulator and a contrast checker, confirming that it meets WCAG Level AA standards.\
